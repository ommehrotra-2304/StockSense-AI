```python
from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

ROOT = Path.cwd().parent if Path.cwd().name == 'notebooks' else Path.cwd()
sales = pd.read_csv(ROOT / 'data/raw/sales.csv', parse_dates=['date'])
sales.head()
```

```python
monthly = sales.assign(month=sales['date'].dt.to_period('M').astype(str)).groupby('month', as_index=False)['units_sold'].sum()
monthly.plot(x='month', y='units_sold', figsize=(12,4), legend=False, title='Monthly Units Sold')
plt.xticks(rotation=90)
plt.tight_layout()
```

```python
top_products = sales.groupby('product_id', as_index=False)['units_sold'].sum().sort_values('units_sold', ascending=False).head(10)
sns.barplot(data=top_products, x='units_sold', y='product_id')
plt.title('Top 10 Products by Units Sold')
plt.tight_layout()
```
